```python
df = pd.read_excel(
    "../../data/Modulo_4_Controle_Vendas.xlsx",
    header=2
)
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
df["Valor Total"] = df["Valor Unitário"] * df["Qtde"]
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
vendas_por_vendedor = df.groupby("Vendedores")["Valor Total"].sum()

print(vendas_por_vendedor)
```

```python
df.groupby(
    ["Vendedores", "Período"]
)["Valor Total"].sum()
```

```python
vendas_vendedor = (
    df.groupby("Vendedores")["Valor Total"]
      .sum()
)

vendas_vendedor.plot(kind="bar")
```

```python
import matplotlib.pyplot as plt

vendas_vendedor = (
    df.groupby("Vendedores")["Valor Total"]
      .sum()
)

print(vendas_vendedor)

vendas_vendedor.plot(kind="bar")

plt.title("Vendas por vendedor")
plt.xlabel("Vendedor")
plt.ylabel("Valor total")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
